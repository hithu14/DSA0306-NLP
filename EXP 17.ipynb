```python
import nltk
from nltk.corpus import wordnet
nltk.download('wordnet')
nltk.download('omw-1.4')
word="car"
synsets=wordnet.synsets(word)
print("word",word)
print("number of synsets",len(synsets))
for synset in synsets:
  print("\nsynset:",synset.name())
  print("definition:",synset.definition())
  print("examples:",synset.examples())
  synonyms=synset.lemmas()
  print("synonyms:",[lemma.name() for lemma in synonyms])
```

```python
import nltk
from nltk.wsd import lesk
from nltk.corpus import wordnet
nltk.download('wordnet')
nltk.download('omw-1.4')
sentence="I went to the bank to deposit money"
words=sentence.split()
result=lesk(words,"bank")
print("sentence",sentence)
print("Ambigous word: bank")
if result:
  print("selected sense",result.name())
  print("definition:",result.definition())
else:
  print("no sense found")
```
