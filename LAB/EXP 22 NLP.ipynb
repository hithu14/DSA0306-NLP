```python
import spacy
nlp = spacy.load("en_core_web_sm")
text = input("Enter a sentence: ")
doc = nlp(text)
nouns = []
print("\nReference Resolution:")
print("---------------------")
for token in doc:
    if token.pos_ in ["NOUN", "PROPN"]:
        nouns.append(token.text)
    if token.pos_ == "PRON":
        if nouns:
            print(token.text, "->", nouns[-1])
        else:
            print(token.text, "-> No reference found")
```
