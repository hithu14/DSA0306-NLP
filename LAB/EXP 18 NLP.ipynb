```python
import re

# Function to check whether an expression is a valid predicate
def is_valid_predicate(expr):
    pattern = r'^[A-Za-z][A-Za-z0-9_]*\([A-Za-z0-9_, ]+\)$'
    return bool(re.match(pattern, expr.strip()))


# Function to parse FOPC expression
def parse_fopc(expression):
    expression = expression.strip()

    # Check logical operators
    if " AND " in expression:
        parts = expression.split(" AND ")
        print("Logical Operator : AND")
        for part in parts:
            parse_fopc(part)

    elif " OR " in expression:
        parts = expression.split(" OR ")
        print("Logical Operator : OR")
        for part in parts:
            parse_fopc(part)

    elif expression.startswith("NOT "):
        print("Logical Operator : NOT")
        parse_fopc(expression[4:])

    # Check quantifiers
    elif expression.startswith("FORALL "):
        parts = expression.split(" ", 2)
        print("Quantifier        :", parts[0])
        print("Variable          :", parts[1])
        parse_fopc(parts[2])

    elif expression.startswith("EXISTS "):
        parts = expression.split(" ", 2)
        print("Quantifier        :", parts[0])
        print("Variable          :", parts[1])
        parse_fopc(parts[2])

    # Check predicate
    elif is_valid_predicate(expression):
        predicate = expression.split("(")[0]
        arguments = expression[expression.find("(")+1:expression.find(")")]

        print("Predicate         :", predicate)
        print("Arguments         :", arguments.split(","))

    else:
        print("Invalid expression")


# Main program
expression = input("Enter FOPC expression: ")

print("\nFOPC Parsing Result")
print("-------------------")

parse_fopc(expression)
```
